```python
import pandas as pd
url = "https://raw.githubusercontent.com/datasets/s-and-p-500/master/data/data.csv"

df = pd.read_csv(url)

df.head()
```

```python
import matplotlib.pyplot as plt
df.plot(x = "Date", y = "SP500")
plt.show()
```

```python
df["change"] = df["SP500"].diff()
df.plot(x="Date", y="change")
plt.show()

```

```python
biggest_gain = df["change"].max()
biggest_drop = df["change"].min()

print(biggest_gain)
print(biggest_drop)
```
